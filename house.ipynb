```python
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing(as_frame=True)
df = data.frame
df.to_numpy()
df.head()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split as ts 

X = df.drop(["MedHouseVal"] , axis=1).values
Y = df["MedHouseVal"].values.reshape(-1 , 1)

# normalisation
X = (X - X.mean(axis = 0))/X.std( axis=0)

# data splitting 
X_train , X_test , Y_train , Y_test = ts(X ,Y , test_size= 0.3 , shuffle= True)


epochs=500 

```

```python
# grad_func(LSSO)

def Lasso(X , Y , alpha = 0.01 , lmd = 0.1):
    m , n = X.shape
    wieghts = np.random.randn( n ,1)
    losses = []

    for i in range(epochs):
        y_pred = X @ wieghts
        error = y_pred - Y

        #lasso grad 

        # Lasso gradient
        grad = (X.T @ error) / m + lmd * np.sign(weights)
        weights -= alpha * grad

        # Lasso loss
        loss = (np.sum(error**2) / (2*m)) + lmd * np.sum(np.abs(weights))
        losses.append(loss)

    return weights, losses
```

```python

# grad_func(RIDGE)

def ridge(X, Y, alpha=0.01, lmd =0.1):
    
    m, n = X.shape
    weights = np.random.randn(n, 1)
    losses = []

    for i in range(epochs):
        y_pred = X @ weights
        error = y_pred - Y

        # Ridge grad
        grad = (X.T @ error) / m + 2 * lmd * weights
        weights -= alpha * grad

        # Ridge loss
        loss = (np.sum(error**2) / (2*m)) + lmd * np.sum(weights**2)
        losses.append(loss)

    return weights, losses
```

```python
# trainning
ridge_weights, ridge_losses = ridge(X_train, Y_train, alpha=0.01, lambda_=0.1, epochs=500)
lasso_weights, lasso_losses = Lasso(X_train, Y_train, alpha=0.01, lambda_=0.1, epochs=500)
```

```python
# Evaluation 
from sklearn.metrics import mean_squared_error

ridge_preds = X_test @ ridge_weights
lasso_preds = X_test @ lasso_weights

ridge_mse = mean_squared_error(Y_test, ridge_preds)
lasso_mse = mean_squared_error(Y_test, lasso_preds)

print("Ridge MSE:", ridge_mse)
print("Lasso MSE:", lasso_mse)
```

```python
# Visualize Loss Convergence

import matplotlib.pyplot as plt
plt.plot(ridge_losses, label="Ridge Loss")
plt.plot(lasso_losses, label="Lasso Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Convergence of Loss Functions")
plt.legend()
plt.show()

# Comparing w
print("Ridge Weights:\n", ridge_weights.flatten())
print("Lasso Weights:\n", lasso_weights.flatten())
```
